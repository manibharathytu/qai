```python
import random 
```

```python
def do_job(n):
    count =0
    for num in range(n):
             tmp = round(random.random()*10)
             if tmp==1:
                     count+=1
    return count 
```

```python
count=0
turn=0
```

```python
n=1000
turns = 10

for i in range(turns):
    turn+=1
    count += do_job(n)
    print(count/turn) 
```
